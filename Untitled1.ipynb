```python
# Cell 1: Import all required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings
warnings.filterwarnings('ignore')

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional, SpatialDropout1D
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc

import time

# Set random seeds for reproducibility
np.random.seed(42)
import tensorflow as tf
tf.random.set_seed(42)

print("✅ All libraries imported successfully!")
print(f"TensorFlow version: {tf.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"Pandas version: {pd.__version__}")
```

```python
# Cell 2: Load the IMDB dataset
df = pd.read_csv('data/IMDB Dataset.csv')

print("="*70)
print("DATASET LOADED")
print("="*70)
print(f"\nDataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")

print(f"\nFirst 3 rows:")
print(df.head(3))

print(f"\nData types:")
print(df.dtypes)

print(f"\nMissing values:")
print(df.isnull().sum())

print(f"\nSentiment distribution:")
print(df['sentiment'].value_counts())

# Check for class balance
positive_count = (df['sentiment'] == 'positive').sum()
negative_count = (df['sentiment'] == 'negative').sum()

print(f"\nPositive reviews: {positive_count}")
print(f"Negative reviews: {negative_count}")
print(f"Dataset is {'balanced' if positive_count == negative_count else 'imbalanced'}")
```

```python
# Cell 3: Define and apply text cleaning function
def clean_text(text):
    """
    Clean text while preserving sentiment information.
    We keep the cleaning light - don't remove stopwords!
    """
    text = text.lower()
    text = re.sub(r'<br />', ' ', text)  # Replace HTML breaks with space
    text = re.sub(r'http\S+|www.\S+', '', text)  # Remove URLs
    text = re.sub(r'[^\w\s]', '', text)  # Remove punctuation
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text

print("="*70)
print("TEXT CLEANING")
print("="*70)
print("\nCleaning all reviews... This will take 1-2 minutes...")

# Apply cleaning
df['review_clean'] = df['review'].apply(clean_text)

print("\nText cleaning completed!")

# Show before/after example
print("\n" + "="*70)
print("EXAMPLE: BEFORE AND AFTER CLEANING")
print("="*70)
sample_idx = 0

print("\nORIGINAL:")
print(df['review'].iloc[sample_idx][:300])

print("\nCLEANED:")
print(df['review_clean'].iloc[sample_idx][:300])

# Check if cleaning worked
print(f"\nTotal reviews cleaned: {len(df)}")
print(f"Average review length (words): {df['review_clean'].apply(lambda x: len(x.split())).mean():.1f}")
```

```python
# Cell 4: Encode sentiment labels
df['label'] = df['sentiment'].map({'positive': 1, 'negative': 0})

print("="*70)
print("LABEL ENCODING")
print("="*70)

print("\nLabels encoded:")
print("  'positive' → 1")
print("  'negative' → 0")

print(f"\nLabel distribution:")
print(df['label'].value_counts().sort_index())

# Show sample
print(f"\nSample data with labels:")
print(df[['sentiment', 'label']].head(10))

# Verify encoding
print(f"\nVerification:")
print(f"  Positive labels: {(df['label'] == 1).sum()}")
print(f"  Negative labels: {(df['label'] == 0).sum()}")
print(f"  Total: {len(df)}")
```

```python
# Cell 5: Split data into train, validation, and test sets

print("="*70)
print("DATA SPLITTING")
print("="*70)

# First split: 80% train+val, 20% test
X_train_val, X_test, y_train_val, y_test = train_test_split(
    df['review_clean'], 
    df['label'], 
    test_size=0.2, 
    random_state=42,
    stratify=df['label']
)

print("\nFirst split (Train+Val vs Test):")
print(f"  Train+Val: {len(X_train_val)} samples (80%)")
print(f"  Test: {len(X_test)} samples (20%)")

# Second split: 80% train, 20% val (from train+val)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, 
    y_train_val,
    test_size=0.2,
    random_state=42,
    stratify=y_train_val
)

print("\nSecond split (Train vs Validation):")
print(f"  Train: {len(X_train)} samples (64% of total)")
print(f"  Validation: {len(X_val)} samples (16% of total)")

print("\n" + "="*70)
print("FINAL DATA SPLIT SUMMARY")
print("="*70)
print(f"\nTraining set: {len(X_train):,} samples")
print(f"Validation set: {len(X_val):,} samples")
print(f"Test set: {len(X_test):,} samples")
print(f"Total: {len(df):,} samples")

# Check class distribution in each set
print("\nClass distribution (Positive vs Negative):")
print(f"  Train: {y_train.value_counts().sort_index().tolist()}")
print(f"  Val: {y_val.value_counts().sort_index().tolist()}")
print(f"  Test: {y_test.value_counts().sort_index().tolist()}")

print("\n✓ All sets are balanced!")
```

```python
# Cell 6: Tokenize text and create padded sequences

# Hyperparameters
VOCAB_SIZE = 15000
MAX_LENGTH = 250
EMBEDDING_DIM = 128

print("="*70)
print("TOKENIZATION & PADDING")
print("="*70)
print(f"\nHyperparameters:")
print(f"  Vocabulary Size: {VOCAB_SIZE:,}")
print(f"  Max Sequence Length: {MAX_LENGTH}")
print(f"  Embedding Dimension: {EMBEDDING_DIM}")

# Create and fit tokenizer on training data only
print("\nCreating tokenizer...")
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token='<OOV>')
tokenizer.fit_on_texts(X_train)

# Get vocabulary info
word_index = tokenizer.word_index
print(f"\nTotal unique words in training data: {len(word_index):,}")
print(f"Using top {VOCAB_SIZE:,} most frequent words")

# Show some examples
print(f"\nTop 20 most frequent words:")
sorted_words = sorted(word_index.items(), key=lambda x: x[1])[:20]
for word, idx in sorted_words:
    print(f"  {word}: {idx}")

# Convert to sequences
print("\nConverting text to sequences...")
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_val_seq = tokenizer.texts_to_sequences(X_val)
X_test_seq = tokenizer.texts_to_sequences(X_test)

# Show example
print(f"\nExample conversion:")
print(f"Original text: {list(X_train)[0][:100]}...")
print(f"Sequence: {X_train_seq[0][:20]}...")

# Pad sequences
print("\nPadding sequences to fixed length...")
X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LENGTH, padding='post', truncating='post')
X_val_pad = pad_sequences(X_val_seq, maxlen=MAX_LENGTH, padding='post', truncating='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=MAX_LENGTH, padding='post', truncating='post')

print("\n" + "="*70)
print("FINAL SHAPES")
print("="*70)
print(f"X_train_pad: {X_train_pad.shape}")
print(f"X_val_pad: {X_val_pad.shape}")
print(f"X_test_pad: {X_test_pad.shape}")

print(f"\nEach sequence has {MAX_LENGTH} numbers representing words")
print("Padding='post' means zeros added at the end")
print("Truncating='post' means long reviews cut from the end")
```

```python
# Cell 10: Rebuild LSTM with optimized settings

def build_optimized_lstm():
    """
    Optimized LSTM - reduced dropout, better learning rate
    """
    model = Sequential([
        # Embedding layer
        Embedding(
            input_dim=VOCAB_SIZE,
            output_dim=EMBEDDING_DIM,
            input_length=MAX_LENGTH,
            name='embedding'
        ),
        
        # Reduced dropout from 0.2 to 0.1
        SpatialDropout1D(0.1),
        
        # First LSTM - reduced dropout
        LSTM(
            128, 
            return_sequences=True,
            dropout=0.1,              # Reduced from 0.2
            recurrent_dropout=0.1     # Reduced from 0.2
        ),
        
        # Second LSTM - reduced dropout
        LSTM(
            64,
            dropout=0.1,              # Reduced from 0.2
            recurrent_dropout=0.1     # Reduced from 0.2
        ),
        
        # Dense layers - reduced dropout
        Dense(64, activation='relu'),
        Dropout(0.3),                 # Reduced from 0.5
        
        Dense(32, activation='relu'),
        Dropout(0.2),                 # Reduced from 0.3
        
        # Output layer
        Dense(1, activation='sigmoid')
    ])
    
    # Build model
    model.build(input_shape=(None, MAX_LENGTH))
    
    # Higher learning rate
    optimizer = Adam(learning_rate=0.001)  # Increased from 0.0005
    
    model.compile(
        optimizer=optimizer,
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    
    return model

print("="*70)
print("REBUILDING LSTM WITH OPTIMIZED SETTINGS")
print("="*70)

model = build_optimized_lstm()

print("\nModel Architecture:")
model.summary()

print("\nKey Changes:")
print("  - Reduced all dropout rates by ~50%")
print("  - Increased learning rate: 0.0005 → 0.001")
print("  - This should improve learning significantly")
```

```python
# Cell 9: Train LSTM Model

print("="*70)
print("TRAINING LSTM MODEL")
print("="*70)
print("\nThis will take approximately 15-20 minutes...")
print("Training on 32,000 samples")
print("Validating on 8,000 samples")
print("\nStarting training...\n")

start_time = time.time()

history = model.fit(
    X_train_pad, y_train,
    batch_size=64,
    epochs=15,
    validation_data=(X_val_pad, y_val),
    callbacks=callbacks,
    verbose=1
)

end_time = time.time()
training_time = (end_time - start_time) / 60

print("\n" + "="*70)
print("TRAINING COMPLETED")
print("="*70)
print(f"\nTotal training time: {training_time:.2f} minutes")
print(f"Epochs completed: {len(history.history['accuracy'])}")

# Print final metrics
final_train_acc = history.history['accuracy'][-1]
final_val_acc = history.history['val_accuracy'][-1]
final_train_loss = history.history['loss'][-1]
final_val_loss = history.history['val_loss'][-1]

print(f"\nFinal Training Accuracy: {final_train_acc:.4f} ({final_train_acc*100:.2f}%)")
print(f"Final Validation Accuracy: {final_val_acc:.4f} ({final_val_acc*100:.2f}%)")
print(f"Final Training Loss: {final_train_loss:.4f}")
print(f"Final Validation Loss: {final_val_loss:.4f}")

print(f"\nBest model saved to: models/best_lstm_model.h5")
```
